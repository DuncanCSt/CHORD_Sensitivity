Function to great layout arrangements

```python
import numpy as np
```

```python
n_x = 4
n_y = 4
top_left = (308839.20, 5466245.45)
bottom_left = (308832.01, 5466173.10)
top_right = (308918.61, 5466230.69)
bottom_right = (308907.52, 5466161.14)

def create_range(
        top_left, bottom_left, top_right, bottom_right,
        n_x, n_y
):

    def create_line(start, end, n):
        row_x = np.linspace(start[0], end[0], n)
        row_y = np.linspace(start[1], end[1], n)
        return [(row_x[i], row_y[i]) for i in range(n)]
    top_row = create_line(top_left, top_right, n_x)
    bottom_row = create_line(bottom_left, bottom_right, n_x)

    locs = []
    for i in range(n_x):
        locs.extend(create_line(top_row[i], bottom_row[i], n_y))

    return locs

np.savetxt("location_A_2.txt", locs, delimiter=" ")
```

## Location A_2

```python
top_left = (308839.20, 5466245.45)
bottom_left = (308832.01, 5466173.10)
top_right = (308918.61, 5466230.69)
bottom_right = (308907.52, 5466161.14)
n_x = 4
n_y = 4

A2 = create_range(top_left=top_left, bottom_left=bottom_left, top_right=top_right, bottom_right=bottom_right,
                  n_x=4, n_y=4)


np.savetxt("location_A_2.txt", A2, delimiter=" ")
```
