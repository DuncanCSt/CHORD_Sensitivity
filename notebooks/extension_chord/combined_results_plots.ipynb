```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
locations[4:8]
```

```python
locations = ['A', 'A2', 'B', 'C', 'D', 'E', 'F', 'G']

folder = 'results_2'

for group in [locations[0:4], locations[4:8]]:
    fig, axes = plt.subplots(
        4, 2, figsize=(11, 16), gridspec_kw={'width_ratios': [1, 1]}
    )
    for i, location in enumerate(group):
        briggs0 = mpimg.imread(f'{folder}/compact_{location}/beam_briggs0.png')
        uniform = mpimg.imread(f'{folder}/compact_{location}/beam_uniform.png')
        sweep = mpimg.imread(f'{folder}/compact_{location}/briggs_sweep.png')

        # axes[i, 0].imshow(briggs0)
        axes[i, 0].imshow(uniform)
        axes[i, 1].imshow(sweep)

        for ax in axes[i]:
            ax.axis('off')

        axes[i, 0].text(
            -0.1, 0.5, location,
            transform=axes[i, 0].transAxes,
            fontsize=30,
            fontweight='bold',
            va='center',
            ha='center'
        )

    axes[0, 0].set_title('Uniform (R= -2)', fontsize=30)
    axes[0, 1].set_title('Briggs sweep', fontsize=30)

    plt.tight_layout()
    plt.savefig(f'../../thesis/plots/ext_chord/v2_combined_results_{group[0]}_{group[-1]}.png', dpi=300)
    plt.show()
```

## Combined locations

```python
locations = ['A', 'A2', 'B', 'C', 'D', 'E', 'F', 'G']
joint_locations = []
for i, location in enumerate(locations):
    for alt_location in locations[i+1:]:
        joint_locations.append(f'{location}_{alt_location}')
joint_locations
```

```python
[joint_locations[j:j+4] for j in range(0, len(joint_locations), 4) if len(joint_locations[j:j+4]) == 4]
```

```python
for group in [joint_locations[j:j+4] for j in range(0, len(joint_locations), 4) if len(joint_locations[j:j+4]) == 4]:
    fig, axes = plt.subplots(
        4, 2, figsize=(11, 16), gridspec_kw={'width_ratios': [1, 1]}
    )
    for i, location in enumerate(group):
        briggs0 = mpimg.imread(f'{folder}/compact_{location}/beam_briggs0.png')
        uniform = mpimg.imread(f'{folder}/compact_{location}/beam_uniform.png')
        sweep = mpimg.imread(f'{folder}/compact_{location}/briggs_sweep.png')

        # axes[i, 0].imshow(briggs0)
        axes[i, 0].imshow(uniform)
        axes[i, 1].imshow(sweep)

        for ax in axes[i]:
            ax.axis('off')

        axes[i, 0].text(
            -0.15, 0.5, location.replace('_', '+'),
            transform=axes[i, 0].transAxes,
            fontsize=30,
            fontweight='bold',
            va='center',
            ha='center'
        )

    axes[0, 0].set_title('Uniform (R= -2)', fontsize=30)
    axes[0, 1].set_title('Briggs sweep', fontsize=30)

    plt.tight_layout()
    plt.savefig(f'../../thesis/plots/ext_chord/v2_combined_results_{group[0]}_{group[-1]}.png', dpi=300)
    plt.show()
```
